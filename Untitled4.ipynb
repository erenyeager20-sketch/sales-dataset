```python
# Step 1: Imports
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
from ipywidgets import interact, widgets
from IPython.display import display

# Step 2: Connect to the SQLite database
conn = sqlite3.connect("sales_data.db")
cursor = conn.cursor()

# Step 3: Create 'sales' table and insert sample data (only first run)
cursor.execute("""
CREATE TABLE IF NOT EXISTS sales (
    id INTEGER PRIMARY KEY,
    product TEXT,
    quantity INTEGER,
    price REAL
)
""")

# Optional: Insert data (comment out after first run)
sample_data = [
    ('Apples', 10, 1.5),
    ('Bananas', 5, 0.8),
    ('Oranges', 8, 1.2),
    ('Apples', 6, 1.5),
    ('Bananas', 7, 0.8),
]
cursor.executemany("INSERT INTO sales (product, quantity, price) VALUES (?, ?, ?)", sample_data)
conn.commit()

# Step 4: Load full summary into DataFrame
query = """
SELECT product, 
       SUM(quantity) AS total_qty, 
       SUM(quantity * price) AS revenue
FROM sales
GROUP BY product
"""
df_summary = pd.read_sql_query(query, conn)

# Step 5: Display the summary
print("Basic Sales Summary:")
print(df_summary)

# Step 6: Plot bar chart of revenue
df_summary.plot(kind='bar', x='product', y='revenue', legend=False, color='skyblue')
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.tight_layout()
plt.show()

# Step 7: Interactivity – Filter sales by product

# Get unique product list
products = pd.read_sql_query("SELECT DISTINCT product FROM sales", conn)['product'].tolist()

def show_product_sales(product):
    query = f"""
    SELECT id, product, quantity, price, (quantity * price) as total_price
    FROM sales
    WHERE product = '{product}'
    """
    df = pd.read_sql_query(query, conn)
    display(df)

interact(show_product_sales, product=widgets.Dropdown(options=products, description='Product:'))
```
